```python
# Define the reflex agent for the two-room vacuum environment
def vacuum_agent(location, room_status):
    # Normalize inputs for safe matching
    loc = location.upper()
    status = room_status.lower()

    # Simple reflex rules
    if status == "dirty":
        return "Suck"
    elif status == "clean":
        # If Room A is clean, move to B. If Room B is clean, move to A.
        if loc == "A":
            return "Move to B"
        elif loc == "B":
            return "Move to A"
    return "Stay Put"

# A chronological list of percepts representing the simulation steps
# Format: (Location, Room Status)
percept_history = [
    ("A", "dirty"),  # Step 1: Starts in Room A, finds it dirty
    ("A", "clean"),  # Step 2: Cleans Room A, checks again
    ("B", "dirty"),  # Step 3: Moves to Room B, finds it dirty
    ("B", "clean")   # Step 4: Cleans Room B, checks again
]

# Run the simulation loop
print("--- Two-Room Vacuum Simulation ---")
print(f"{'Step':<6} | {'Percept (Room, Status)':<25} | {'Agent Action':<15}")
print("-" * 52)

for index, percept in enumerate(percept_history, 1):
    room, status = percept
    action = vacuum_agent(room, status)
    print(f"Step {index:<1} | ({room}, {status:<5}){' ':<15} | {action}")
```
